Consider an electron, $m_e=9.109\times10^{-31}$ kg, in a box of length $L = 10^{-8}$ m.  Suppose that at time $t=0$ the wavefunction of the electron has the form:
$$
\psi(x,0) = \exp \biggl[ -\frac{(x-x_0)^2}{2\sigma^2} \biggr]
            e^{i\kappa x},
$$
where
$$
x_0 = {L\over2},\qquad
\sigma = 1\times10^{-10}\,\mathrm{m},\qquad
\kappa = 5\times10^{10}\,\mathrm{m}^{-1},
$$
