```{exercise} Euler's Method
:label: icc_6-1
:number: 1

Consider the first-order non-linear ODE, which is an [Abel equation of the first kind](https://en.wikipedia.org/wiki/Abel_equation_of_the_first_kind), for which there are not straightforward analytic solutions:

$$ \frac{dx}{dt} = f(x,t)  = -x^3 + sin(t) $$

Let's solve it numerically using the Euler method.

+ Write a function called `Euler` which takes a function $f(x,t)$ and solves for $x(t)$ on the interval $t = t_0$ to $t=t_f$, given some $x(t_0) = x_0$ using the Euler method to advance iteratively through time. The function should take $N$, the total number of timesteps as an input parameter.
+ Use your function to solve $x(t)$ on the interval $t=0$ to $t=10$ with N= [10,25,50] timesteps. 
+ Plot your results for the three solutions on the same figure. 

```


````python
import numpy as np
import matplotlib.pyplot as plt

def f(x,t):
    return -x**3 + np.sin(t)

t0 = 0.0           # initial interval
x0 = 0.0           # Initial condition
tf = 10.0          # final time to integrate to
Nsteps = 1000      # number of time steps


def Euler(f,t0,tf,x0,N):
    """ Perform a forward euler solution of an ODE f = dx/dt from t0 to tf over N timesteps
    given ivp value x0 at t0. 
    return array of all timesteps taken  and values of x(t) across the timesteps"""
    t_vals = [t0]
    x_vals = [x0]
    return t_vals, x_vals

fig, ax = plt.subplots(1)
ax.plot(*Euler(f,t0,tf,x0,10), marker='.',label='N=10')
ax.plot(*Euler(f,t0,tf,x0,25), marker='.',label='N=25')
ax.plot(*Euler(f,t0,tf,x0,50), marker='.',label='N=50')
ax.legend()
ax.set_xlabel('$t$')
ax.set_ylabel('$x(t)$')
ax.set_title('Euler solution of $dx/dt = -x^3 + \sin(t), \ x(0) = 0$')
````

````python
# imports
from math import pi,sqrt
import numpy as np
import matplotlib.pyplot as plt

# units
GM = 4 * pi**2 

class State:
    def __init__(self, x, y, u, v):
        self.x = x
        self.y = y
        self.u = u
        self.v = v

    def __str__(self):
        return f"x ={self.x:10.6f} y ={self.y:10.6f} u ={self.u:10.6f} v ={self.v:10.6f}"

    def rhs(self):
        """
        computes and returns the right hand side of the ODEs for xdot, ydot, udot, and vdot
        """
        # current magnitude of radius vector
        r = sqrt(self.x**2 + self.y**2)

        # position ODE
        xdot = self.u
        ydot = self.v

        # velocity ODE
        udot = -GM*self.x/r**3
        vdot = -GM*self.y/r**3
        
        return xdot, ydot, udot, vdot

    def add_(self,other):
        """ adds the state vector to an other state vector, and sets the sum as the new state """
        self.x += other.x
        self.y += other.y
        self.u += other.u
        self.v += other.v

    def mult_(self, value):
        """ multiplies the state vector by a scalar value and sets the product as the new state """
        self.x *= value
        self.y *= value
        self.u *= value
        self.v *= value
 

def Euler_trajectory(state0, dt, tf=1):
    t = 0.0
    x_vals = [state0.x]
    y_vals = [state0.y]
    t_vals = [0.0]
    #####
    return t_vals, x_vals, y_vals
        

fig,ax = plt.subplots(1)
for h in [0.1,0.05,0.025,0.00125]:
    state0 = State(0,1,-sqrt(GM/1),0)
    t,x,y = Euler_trajectory(state0,h,1)
    ax.plot(x,y,label="dt = {}".format(h))

# mark the Sun
ax.scatter([0], [0], s=250, marker=(20, 1), color="k")
ax.scatter([0], [0], s=200, marker=(20, 1), color="y")

# draw the analytic solution
theta = np.linspace(0.0, 2.0*np.pi, 360)
ax.plot(np.cos(theta), np.sin(theta), ls=':', color='k',label='exact')

# labels
ax.set_xlabel('x [au]')
ax.set_ylabel('y [au]')
ax.set_title('Orbit with Forward Euler')
ax.legend()

ax.axis('equal')
````
