## Boundary Value Problems
Boundary Value Problems (BVPs) are a class of ODE problems for which we specify the behavior of the unknown function at two points in the domain, rather than the behavior of all our functions at the same (initial) point, as we do in IVPs. 

```{note} Nomenclature Conventions
:class: dropdown
Typically:
+ IVPs are posed assuming the independent variable you are iterating across is time 
+ BVPs are posed assuming the independent varaible you are iterating across is space

This is largely due to the frequency of the types of physics problems that have these forms, but the actual distinction is in terms of at how many points on the domain the constraints are specified not the type of independent variable. 

For example:
Take a kinematic trajectory problem for $x(t)$ with two ODEs: $dx/dt = v(t); dv/dt = a(t)$
+ If we have $x(0)$, and $v(0)$, it is an IVP. 
+ If we have $x(0)$ and $x(t_1)$, it is a BVP.
```

For example, let's consider a Poisson equation on the domain $x = [0,1]$
$$ y'' = f(x,y) $$
for which we want to solve for $y(x)$ and we the boundary values $y(0) = a$ and $y(1) = b$.

We already know what to do first! This is a second order ODE, so we need to turn it into two first-order ODEs:
$$ y' = z(x,y); \ z' = f(x,y) $$

We know our boundary conditions:
$y(0) = a, y(1) = b$

And we know that now, we could solve this as an IVP using the methods we've already talked about, except we need a value for the boundary condition in $z$, so let's call it, for now:
$$z(0) = \eta$$

We now need a way to figure out $\eta$:

### Shooting Method
So, what if we just trial and error a guess for $\eta$? If we guess $\eta$ correctly, we'll know because it will be the one that makes the solution satisfy the other BC: $y(1)=b$. 

```{note} Shooting Method Algorithm
1. Make a guess for $\eta$
2. Solve the IVP using $z(0) = \eta$ for the corresponding solution $y_\eta(x)$
3. Iterate 1 & 2 using a root-finding method to determine the solution for which $b - y_\eta(1) = 0$.
```

### Example: Projectile Motion
Let's consider projectile motion, the equation of motion for which is a second-order ODE:
$$ \frac{d^2{\bf x}}{dt^2} = -{\bf a}$$
where ${\bf x} = x,y$ and ${\bf a} = [0,-g]$

Which can be written as two familiar first order ODEs:

$$ \dot{\bf{v}} = [0,-g]; \ \  \dot{{\bf x}} = {\bf v} $$

(The exact solutions for which are given to us by the kinematic equations)

We can write the RHS of the ODE function as:

````python
def rhs(x,y,u,v):
    g = 9.81
    xdot = u
    ydot = v
    udot = 0.0
    vdot = -g
    return np.array([xdot, ydot, udot, vdot])
````

### Eigenvalue Problems
Eigenvalue problems are a special type of boundary value problems where the ODEs are **linear and homogenous**:    
every term in the ODE is linear in the dependent variable

#### Ex: Infinite Square Well
Consider the time-independent Schrodinger equation, where every term is linear in $\psi(x)$: 
$$-\frac{\hbar}{2\pi} \frac{d^{2}\psi}{dx^{2}} + V(x)\psi(x) = E\psi(x)$$

for particle in a square potential well with infinitely high walls: 
\begin{align}
V(x) &= 0 \quad \text{for } 0 < x < L \\ V(x) &= \infty \quad \text{elsewhere}
\end{align}

Recall that $\psi(x)$ is a type of probability distribution function and these conditions tell us the particle can't exist at the boundaries. Thus,the corresponding BCs are:
$$ \psi(0) = 0 \\ \psi(L) = 0 $$

This second order ODE can once again be similarly posed as two first order ODEs:
$$\frac{d\psi}{dx} = \phi(x) \\ \frac{d\phi}{dx} = \frac{2m}{\hbar^2}[V(x) - E]\psi(x)$$

However, looking for the initial conditions of the derivative $\phi(0)$ is not helpful here, as quantum mechanics tells us that these solutions only exist for eigenvalues of the energy $E$. Because the Schrodinger equation is linear in $\psi(x)$, there is no unique general form solution to be found by iterating in $\phi$. 

For an eigenvalue problem, we can instead modify the shooting method to look for eigenvalues of $E$ that also satisfy the boundary conditions in $\psi$.
